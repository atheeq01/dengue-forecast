### Imports

```python
from __future__ import annotations

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from src.config import COMBINED_SILVER_PATH

pd.set_option("display.max_columns", 200)
pd.set_option("display.max_rows", 100)
pd.set_option("display.width", 200)
```

### Load Sprint 1 data

```python
df = pd.read_parquet(COMBINED_SILVER_PATH)

print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

df
```

### Basic ordering

```python
df["week_start"] = pd.to_datetime(
    df["week_start"],
    errors="coerce",
)

df = df.sort_values(
    ["district", "week_start"]
).reset_index(drop=True)

print(
    "Sorted:",
    df[["district", "week_start"]]
    .head(20)
)
```

### Verify one row per district/week

```python
duplicates = df[
    df.duplicated(
        subset=["district", "week_start"],
        keep=False,
    )
]

print("Duplicate rows:", len(duplicates))

if not duplicates.empty:
    print(duplicates.head(50))
```
